```python
from logitdiff_widget import (
    load_model_and_tokenizer,
    build_single_model_widget,
    build_prompt_logitdiff_widget,
    build_generation_logitdiff_widget,
    build_prompt_payload_widget,
    build_generation_payload_widget,
)
```

## Load a model once

Change the model name below if needed. The loaded model/tokenizer pair is then reused by the `Logit Lens` widget.


```python
MODEL_NAME = "meta-llama/Llama-3.2-1B"
DEVICE_MAP = "auto"  # or "cpu"

model, tokenizer = load_model_and_tokenizer(
    MODEL_NAME,
    device_map=DEVICE_MAP,
)
```

## Load a comparison model

Use a second related model here for live pairwise `Prompt LogitDiff` and `Gen LogitDiff`.


```python
COMPARISON_MODEL_NAME = MODEL_NAME
COMPARISON_DEVICE_MAP = DEVICE_MAP

comparison_model, comparison_tokenizer = load_model_and_tokenizer(
    COMPARISON_MODEL_NAME,
    device_map=COMPARISON_DEVICE_MAP,
)
```

## Logit Lens

This widget runs live on the loaded model. You can set the prompt, readout (`ModelNorm` / `Raw`), metric, `force_include_input`, `force_include_output`, start/end index, layer collapse, and marginals.


```python
build_single_model_widget(
    model,
    tokenizer,
    default_prompt="The capital of France is",
)
```

## Prompt LogitDiff

This widget runs live on the two loaded models and exposes the main prompt-side controls directly.


```python
build_prompt_logitdiff_widget(
    model,
    tokenizer,
    comparison_model,
    comparison_tokenizer,
    default_prompt="The capital of France is",
)
```

## Gen LogitDiff

This widget uses real `model.generate(...)` on the two loaded models, then plots the generation-side diff with the same heatmap controls.


```python
build_generation_logitdiff_widget(
    model,
    tokenizer,
    comparison_model,
    comparison_tokenizer,
    default_prompt="The capital of France is",
)
```

## Saved payload viewers

If you already have cached prompt-side or generation-side payloads, you can still open them directly with the lighter viewers below.


```python
build_prompt_payload_widget(payload_path="")
```
